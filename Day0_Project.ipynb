### Step 1: Initialize the client

```python
from qdrant_client import QdrantClient, models
from google.colab import userdata

client = QdrantClient("https://fa13d3f3-6758-4e1a-9777-f66dc4f300c2.us-east4-0.gcp.cloud.qdrant.io", api_key=userdata.get('api-key'))
```

### Step 2: Create the collection

```python
collection_name = "day0_first_system"

client.create_collection(
    collection_name=collection_name,
    vectors_config=models.VectorParams(
        size=4,
        distance=models.Distance.COSINE
    )
)

# create a payload index for enabling filtering
client.create_payload_index(
    collection_name=collection_name,
    field_name="category",
    field_schema=models.PayloadSchemaType.KEYWORD
)
```

### Step 3: Insert Points

```python
points = [
    models.PointStruct(
        id=1,
        vector=[0.9, 0.1, 0.1, 0.8], # High affordability, High innovation
        payload={
            "name": "Budget smartphone",
            "category": "electronics",
            "price": 299
        },
    ),
    models.PointStruct(
        id=2,
        vector=[0.2, 0.9, 0.8, 0.5], # High quality, High popularity
        payload={
            "name": "BestSelling Novel",
            "Category": "books",
            "price": 19
        },
    ),
    models.PointStruct(
        id=3,
        vector=[0.8, 0.2, 0.3, 0.9], # High affordability, High innovation
        payload={
            "name": "Smart cleaning robot",
            "Category": "electronics",
            "price": 670
        },
    ),
    models.PointStruct(
        id=4,
        vector=[0.9, 0.9, 0.3, 0.1], # High affordability, High quality
        payload={
            "name": "Comfortable chair",
            "Category": "House holds",
            "price": 250
        },
    ),
    models.PointStruct(
        id=5,
        vector=[0.8, 0.2, 0.9, 0.3], # High affordability, High popularity
        payload={
            "name": "Smart car",
            "Category": "vehicles",
            "price": 1200
        },
    ),
    models.PointStruct(
        id=6,
        vector=[0.3, 0.9, 0.3, 0.9], # High quality, High innovation
        payload={
            "name": "Nano raincoat",
            "Category": "Clothes",
            "price": 50
        },
    ),
]

client.upsert(collection_name=collection_name, points=points)
```

### Step 5: Test Searches

```python
# Define the query vector for "Affordable and innovative"
query_vector = [0.85, 0.4, 0.2, 0.89]

# Basic similarity search
basic_results = client.query_points(
    collection_name=collection_name,
    query=query_vector
)

# Filtered Search
filtered_results = client.query_points(
    collection_name=collection_name,
    query=query_vector,
    query_filter=models.Filter(
        must=[
            models.FieldCondition(
                key="category",
                match=models.MatchValue(value="electronics")
            )
        ]
    )
)

print("Filtered search results: ", filtered_results)
```
